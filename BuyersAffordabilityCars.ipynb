Exploratory Data Analysis

Vehicle Type Count


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


sns.countplot(x='Make', data=df)
plt.title('Vehicle Type Distribution')
plt.show()
```

Salary vs Car Type

```python
sns.boxplot(x='Make', y='Total_salary', data=df)
plt.title('Salary vs Vehicle Type')
plt.show()
```

Age distribution

```python
sns.histplot(df['Age'], bins=5, kde=True)
plt.title('Age Distribution')
plt.show()
```

Gender Preference Analysis

```python
sns.countplot(x='Make', hue='Gender', data=df)
plt.title("Gender Preference by Car Type")
plt.show()
```

House loan impact on purchasing

```python
sns.countplot(x='Make', hue='House_loan', data=df)
plt.title("House Loan Impact")
plt.show()
```

Partner salary impact

```python
sns.countplot(x='Partner_working', hue='Make', data=df)
plt.title("  Car Type by Partner salary")
plt.show()
```

Correlation Heatmap

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
